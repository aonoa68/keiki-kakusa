## 3. ローレンツ曲線とは？

ローレンツ曲線は、所得や資産の分布の不平等度を視覚化するグラフです。

### 作成手順
1. 世帯を所得（または資産）の低い順に並べる
2. **横軸**: 累積世帯比率（下位何%の世帯か）
3. **縦軸**: 累積所得比率（その世帯が全体の所得の何%を占めるか）

### 読み方
- **完全平等線（45度線）**: 全員が同じ所得を持つ場合の線
- **ローレンツ曲線が45度線から離れるほど**: 格差が大きい

例えば、「下位50%の世帯が、全体の所得の20%しか持っていない」場合、
ローレンツ曲線上の点は (0.5, 0.2) となります。
